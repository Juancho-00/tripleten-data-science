Sustituye los valores ausentes en las columnas `'track'`, `'artist'` y `'genre'` con el string `'unknown'`.

Vamos a examinar la lista para identificar **duplicados implícitos** del género `hiphop`, es decir, nombres mal escritos o variantes que hacen referencia al mismo género musical.

Los duplicados que encontrarás son:

- `hip`
- `hop`
- `hip-hop`

Para solucionarlo, vamos a crear una función llamada `replace_wrong_values()`.


1. Define una función llamada `replace_wrong_values()` que reciba los siguientes parámetros:

- `df`: el DataFrame a modificar
- `column`: el nombre de la columna a trabajar
- `wrong_values`: una lista con los valores incorrectos
- `correct_value`: el valor correcto para reemplazar

2. Dentro de la función, usa un bucle `for` para iterar sobre cada valor incorrecto y aplicar `.replace()`.


```python
def replace_wrong_values(df, column, wrong_values, correct_value):

    for col in wrong_values:
        df["genre"] = df["genre"].replace(col, correct_value)
```

```python
column = 'genre'
wrong_values = ["hip", "hop", "hip-hop"]
correct_value = "hiphop"

replace_wrong_values(df, column, wrong_values, correct_value)

print(
    f'El valor {correct_value} tiene: {df['genre'].value_counts()['hiphop']} registros')
```

Ahora vamos a combinar dos criterios: día y ciudad.

Crea una función llamada `number_tracks()` que reciba dos parámetros:

- `day`: un día de la semana (por ejemplo, `'Monday'`)
- `city`: el nombre de una ciudad (por ejemplo, `'Springfield'`)

Dentro de la función:

1. Filtra el DataFrame por el día.
2. Luego, filtra por la ciudad.
3. Cuenta cuántas veces aparece `'user_id'` en ese filtro.
4. Devuelve ese número como resultado.


```python
def number_tracks(day, city):

    new_df = df[df["day"] == day]
    new_df = new_df[new_df["city"] == city]

    conteo = new_df["user_id"].count()

    return (conteo)
```

```python
print(
    f'La cantidad de usuarios en la ciudad de Springfield el día Monday escucharon {number_tracks("Monday", "Springfield"):,} tracks')

print(
    f'La cantidad de usuarios en la ciudad de Springfield el día Friday escucharon {number_tracks("Friday", "Springfield"):,} tracks')

print(
    f'La cantidad de usuarios en la ciudad de Shelbyville el día Monday escucharon {number_tracks("Monday", "Shelbyville"):,} tracks')

print(
    f'La cantidad de usuarios en la ciudad de Shelbyville el día Friday escucharon {number_tracks("Friday", "Shelbyville"):,} tracks')
```
